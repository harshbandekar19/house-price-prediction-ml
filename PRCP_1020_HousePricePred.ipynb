```python
# Select important numerical columns for outlier analysis
outlier_columns = [
    "SalePrice",
    "LotArea",
    "GrLivArea",
    "TotalHouseSF",
    "TotalOutdoorAreaSF"
]

outlier_results = []

for col in outlier_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ).sum()

    outlier_results.append({
        "Feature": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Outlier_Count": outlier_count,
        "Outlier_Percentage": (
            outlier_count / len(df)
        ) * 100
    })

outlier_summary = pd.DataFrame(outlier_results)

display(outlier_summary.round(2))
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split

# Reload the original dataset
model_df = pd.read_csv(os.path.join(DATA_PATH, "data.csv"))

# Step 1: Represent genuinely absent features explicitly
none_columns = [
    "Alley", "MasVnrType",
    "BsmtQual", "BsmtCond", "BsmtExposure",
    "BsmtFinType1", "BsmtFinType2",
    "FireplaceQu",
    "GarageType", "GarageFinish", "GarageQual", "GarageCond",
    "PoolQC", "Fence", "MiscFeature"
]

model_df[none_columns] = model_df[none_columns].fillna("None")

# Step 2: Correct known structural absences
no_veneer = model_df["MasVnrType"].eq("None")
model_df.loc[no_veneer, "MasVnrArea"] = (
    model_df.loc[no_veneer, "MasVnrArea"].fillna(0)
)

no_garage = model_df["GarageCars"].eq(0)
model_df.loc[no_garage, "GarageYrBlt"] = 0

# Step 3: Recreate engineered features
model_df["HouseAgeAtSale"] = (
    model_df["YrSold"] - model_df["YearBuilt"]
)

model_df["YearsSinceRemodel"] = (
    model_df["YrSold"] - model_df["YearRemodAdd"]
)

invalid_remodel_date = (
    model_df["YearRemodAdd"] > model_df["YrSold"]
)

model_df.loc[
    invalid_remodel_date, "YearsSinceRemodel"
] = model_df.loc[
    invalid_remodel_date, "HouseAgeAtSale"
]

porch_columns = [
    "WoodDeckSF", "OpenPorchSF", "EnclosedPorch",
    "3SsnPorch", "ScreenPorch"
]

model_df["TotalOutdoorAreaSF"] = model_df[porch_columns].sum(axis=1)

model_df["TotalHouseSF"] = (
    model_df["GrLivArea"] + model_df["TotalBsmtSF"]
)

model_df["TotalBathrooms"] = (
    model_df["FullBath"]
    + 0.5 * model_df["HalfBath"]
    + model_df["BsmtFullBath"]
    + 0.5 * model_df["BsmtHalfBath"]
)

# Step 4: Separate input features (X) and target (y)
X = model_df.drop(columns=["SalePrice", "Id"])
y = model_df["SalePrice"]

# Step 5: Split into training (80%) and testing (20%)
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Verify the split
print("Full dataset:", model_df.shape)
print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)

print("\nTraining price summary:")
display(y_train.describe())

print("\nTesting price summary:")
display(y_test.describe())
```

```python
# Identify numerical and categorical columns
numeric_features = X_train.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

# MSSubClass contains numeric codes for building classes,
# so treat it as categorical rather than continuous.
if "MSSubClass" in numeric_features:
    numeric_features.remove("MSSubClass")
    categorical_features.append("MSSubClass")

# Sort for easier inspection
categorical_features = sorted(categorical_features)

print("Number of numerical features:", len(numeric_features))
print("Number of categorical features:", len(categorical_features))

print("\nNumerical features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)

# Check missing values in the training set
print("\nTraining columns with missing values:")
display(
    X_train.isnull()
    .sum()
    .loc[lambda s: s > 0]
    .sort_values(ascending=False)
)
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# 1. Numerical preprocessing
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# 2. Categorical preprocessing
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])

# 3. Apply each preprocessing path to its respective columns
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

# 4. Learn transformations using training data only
X_train_processed = preprocessor.fit_transform(X_train)

# 5. Apply the learned transformations to test data
X_test_processed = preprocessor.transform(X_test)

# 6. Verify the results
print("Original training shape:", X_train.shape)
print("Original testing shape:", X_test.shape)

print("\nProcessed training shape:", X_train_processed.shape)
print("Processed testing shape:", X_test_processed.shape)

print(
    "\nNumber of processed features:",
    len(preprocessor.get_feature_names_out())
)

print(
    "\nMissing values after preprocessing:",
    X_train_processed.shape[0] * 0
    + __import__("numpy").isnan(X_train_processed).sum()
)
```

```python
from sklearn.model_selection import KFold, cross_validate
from sklearn.pipeline import Pipeline
from sklearn.base import clone
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor

# Five-fold cross-validation for regression
cv = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Algorithms with the initial settings we have tested
models = {
    "Linear Regression": LinearRegression(),
    "Ridge Regression": Ridge(alpha=10.0),
    "Lasso Regression": Lasso(alpha=100.0, max_iter=50000),
    "Random Forest": RandomForestRegressor(
        n_estimators=300,
        random_state=42,
        n_jobs=-1
    ),
    "Gradient Boosting": GradientBoostingRegressor(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=3,
        random_state=42
    )
}

# Use metrics with interpretable positive error values
scoring = {
    "MAE": "neg_mean_absolute_error",
    "RMSE": "neg_root_mean_squared_error",
    "R2": "r2"
}

cv_results = []

for name, model in models.items():

    # Every fold learns preprocessing only from its training portion
    pipeline = Pipeline(steps=[
        ("preprocessor", clone(preprocessor)),
        ("model", model)
    ])

    scores = cross_validate(
        pipeline,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring,
        n_jobs=1,
        return_train_score=False
    )

    cv_results.append({
        "Model": name,
        "Mean_CV_MAE": -scores["test_MAE"].mean(),
        "Std_CV_MAE": scores["test_MAE"].std(),
        "Mean_CV_RMSE": -scores["test_RMSE"].mean(),
        "Std_CV_RMSE": scores["test_RMSE"].std(),
        "Mean_CV_R2": scores["test_R2"].mean(),
        "Std_CV_R2": scores["test_R2"].std()
    })

cv_comparison = pd.DataFrame(cv_results).sort_values(
    "Mean_CV_MAE",
    ascending=True
)

display(cv_comparison.round(4))
```

```python
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.model_selection import RandomizedSearchCV
from sklearn.pipeline import Pipeline
from sklearn.base import clone
import pandas as pd

# 1. Combine preprocessing and Gradient Boosting
gb_pipeline = Pipeline(steps=[
    ("preprocessor", clone(preprocessor)),
    ("model", GradientBoostingRegressor(random_state=42))
])

# 2. Define candidate hyperparameter values
param_distributions = {
    "model__n_estimators": [100, 200, 300, 400],
    "model__learning_rate": [0.03, 0.05, 0.08, 0.1],
    "model__max_depth": [2, 3, 4],
    "model__min_samples_leaf": [1, 2, 4, 6],
    "model__subsample": [0.8, 0.9, 1.0]
}

# 3. Set up randomized search
gb_search = RandomizedSearchCV(
    estimator=gb_pipeline,
    param_distributions=param_distributions,
    n_iter=25,
    scoring="neg_mean_absolute_error",
    cv=cv,
    random_state=42,
    n_jobs=-1,
    verbose=1,
    return_train_score=True
)

# 4. Fit the search using training data only
gb_search.fit(X_train, y_train)

# 5. Display the best settings and cross-validation score
print("\nBest hyperparameters:")
print(gb_search.best_params_)

print(
    "\nBest mean CV MAE:",
    f"${-gb_search.best_score_:,.2f}"
)

print("\nBest estimator:")
print(gb_search.best_estimator_)
```

# Challenges Faced Report — House Price Prediction

## 1. Missing Values

**Challenge:** The original dataset contained missing values in 19 columns, including `PoolQC`, `Alley`, `Fence`, `LotFrontage`, `MasVnrArea`, and garage and basement features.

**Technique used:** We investigated feature meanings before choosing a replacement method. Missing categorical values representing absent features were filled with `"None"`. Missing `MasVnrArea` values were set to zero where `MasVnrType` indicated no veneer. Missing garage construction years for houses without garages were represented by zero. For the modeling pipeline, numerical missing values were imputed with the training-set median and categorical missing values with the training-set most frequent category.

**Reason:** Missing values can prevent models from processing data correctly. Distinguishing an absent feature from an unknown measurement preserves useful information and avoids misleading replacements.

## 2. Structural Missingness

**Challenge:** Some missing values represented features that did not exist, rather than information that had simply not been recorded.

**Technique used:** We checked garage and basement indicators. All 81 houses missing garage information had `GarageCars = 0`. Missing basement-quality values matched `TotalBsmtSF = 0` in the records examined.

**Reason:** Replacing these values with `"None"` preserves the distinction between a house without a garage or basement and a house whose feature exists but is unknown.

## 3. Outliers and Unusual Records

**Challenge:** The IQR method flagged 61 sale prices, 69 lot areas, 31 living areas, 28 combined house areas, and 33 combined outdoor areas as potential outliers.

**Technique used:** We calculated the interquartile range and its standard lower and upper boundaries. We then inspected unusual records using related features, including living area, quality, year built, and neighborhood.

**Reason:** Unusual values can represent legitimate luxury homes or large plots, so deleting them automatically could remove important information.

Two records, IDs 524 and 1299, had living areas above 4,000 sq. ft. but recorded sale prices below $300,000. We flagged these records for further review rather than automatically removing or correcting them.

## 4. Skewed Sale Price Distribution

**Challenge:** The original `SalePrice` distribution was positively skewed, with a skewness value of approximately 1.883.

**Technique used:** We created a log-transformed target using `np.log1p()` and compared it with the original target.

**Reason:** A logarithmic transformation compresses large values and can help some regression models learn a more balanced relationship.

The transformed target's skewness decreased to approximately 0.121. However, the log-target model did not outperform the original-target model on every evaluation metric, so we retained both versions for comparison.

## 5. Feature Engineering and Temporal Inconsistency

**Challenge:** Several useful property characteristics were spread across multiple columns. One record also had a remodeling year later than its recorded sale year.

**Technique used:** We engineered `HouseAgeAtSale`, `YearsSinceRemodel`, `TotalOutdoorAreaSF`, `TotalHouseSF`, and `TotalBathrooms`. For House ID 524, the negative `YearsSinceRemodel` value was replaced with `HouseAgeAtSale` as a documented assumption for that inconsistent record; the original date fields were preserved.

**Reason:** Engineered features provide the model with more direct representations of property age, combined area, and bathroom capacity. Investigating the inconsistent date prevents the engineered feature from retaining an unexplained negative value.

The chosen replacement is an assumption rather than a verified correction to the original record.

## 6. Correlated and Redundant Features

**Challenge:** Some engineered features overlapped with their original components. For example, `TotalHouseSF` combines `GrLivArea` and `TotalBsmtSF`, while `TotalBathrooms` combines the individual bathroom counts.

**Technique used:** We compared Linear Regression with Ridge and Lasso regularization.

**Reason:** Strongly correlated features can make ordinary least-squares coefficients unstable. Ridge penalizes squared coefficient values, while Lasso penalizes absolute coefficient values and can set some coefficients to zero.

## 7. Unrealistic Linear Regression Predictions

**Challenge:** Linear Regression produced one negative test-set prediction of approximately −$949,037.68 for House ID 1171, whose actual recorded sale price was $171,000. Some fitted coefficients were also extremely large.

**Technique used:** We inspected prediction errors and coefficients, then evaluated Ridge and Lasso Regression.

**Reason:** Ordinary Linear Regression does not constrain predicted prices to be non-negative. Regularization discourages excessively large coefficients and can improve numerical stability.

In our initial comparison, Ridge and Lasso eliminated negative test predictions and substantially improved RMSE and R² compared with ordinary Linear Regression.

## 8. Mixed Numerical and Categorical Features

**Challenge:** The dataset contained 84 input columns, including continuous measurements, integer-coded classes, and text categories. One-hot encoding expanded the data to 319 processed features.

**Technique used:** We used a `ColumnTransformer` and separate preprocessing pipelines. Numerical columns used median imputation and standard scaling; categorical columns used most-frequent imputation and one-hot encoding.

**Reason:** Most linear models require numerical inputs, and differently scaled measurements can affect regularized regression. One-hot encoding represents categories without imposing arbitrary numerical distances.

`MSSubClass` was treated as categorical because its numerical values represent building classes rather than a continuous measurement.

## 9. Data Leakage Prevention

**Challenge:** Imputation and encoding can leak information if their parameters are learned using validation or test data.

**Technique used:** For model comparison and tuning, we placed preprocessing inside a scikit-learn pipeline and fitted it within each cross-validation training fold.

**Reason:** The validation fold must be treated as unseen data. Learning medians, category mappings, or scaling parameters from that fold would make evaluation less independent.

Deterministic feature calculations use existing property columns, while learned preprocessing statistics are estimated from training data only.

## 10. Model Selection and Hyperparameter Tuning

**Challenge:** Different regression algorithms produced substantially different errors. The initial Linear Regression model performed much worse than the regularized and tree-based models.

**Technique used:** We compared Linear Regression, Ridge, Lasso, Random Forest, and Gradient Boosting using MAE, RMSE, R², and five-fold cross-validation. We used `RandomizedSearchCV` to investigate Gradient Boosting hyperparameters.

**Reason:** A single metric or one split may not reliably identify the best model. Cross-validation provides a broader assessment, while tuning searches for settings that perform well on training-data validation folds.

The tuned Gradient Boosting model achieved a mean cross-validation MAE of approximately $14,882.55 and became our primary model candidate.

## 11. Different Model Evaluation Metrics

**Challenge:** A model can perform differently depending on whether we measure typical absolute error or penalize larger errors more heavily.

**Technique used:** We reported MAE, RMSE, and R² together.

**Reason:** MAE provides an interpretable average absolute error in dollars. RMSE penalizes larger errors more strongly, while R² measures performance relative to a mean-prediction baseline. Using multiple metrics provides a more complete assessment of regression performance.

## 12. Error Variation Across Price Ranges

**Challenge:** The tuned Gradient Boosting model performed differently across price groups. The test-set MAE was approximately $9,645 for houses sold between $100,000 and $150,000, versus approximately $37,375 for houses sold above $300,000.

**Technique used:** We analyzed residuals, plotted actual versus predicted prices, and calculated error metrics by actual-price bands.

**Reason:** Overall metrics can hide differences between lower-priced and more expensive properties. Segment-based evaluation helps reveal potential weaknesses that may matter to particular buyers.

These segments are descriptive findings from the evaluated test split, not a guarantee of future performance.

## 13. Feature Explainability

**Challenge:** Strong prediction performance alone does not explain which features influenced the estimated price.

**Technique used:** We used built-in Gradient Boosting feature importance, permutation importance, and SHAP summary and waterfall plots.

**Reason:** These methods help explain model behavior globally and for individual predictions. They highlighted `TotalHouseSF` and `OverallQual` as important features, along with house age, bathroom capacity, lot area, garage features, and neighborhood.

Importance values and SHAP contributions describe the fitted model; they do not establish that a feature causes a particular market-price change.

## 14. Buyer Recommendation Limitations

**Challenge:** Filtering houses according to a buyer's budget and requirements can return unusual records or no matching properties.

**Technique used:** We implemented rule-based filtering, ranked results by price, quality, or living area, and added a review flag for unusually large houses with comparatively low recorded prices.

**Reason:** Transparent rules allow users to understand why a property matches their preferences. Review flags help identify candidates requiring additional verification without automatically treating them as erroneous.

The current recommendations use historical Ames, Iowa sale records. They do not represent verified current listings or guarantee future market values.

## 15. Test-Set Reuse

**Challenge:** We repeatedly examined test-set metrics while comparing and improving models. This reduces the independence of that test set as a final evaluation benchmark.

**Technique used:** We used training-set cross-validation for model comparison and hyperparameter selection. The existing test metrics are reported transparently as exploratory holdout results.

**Reason:** Repeated decisions based on test-set performance can lead to overly optimistic expectations. A future production assessment should use a genuinely independent dataset or a newly collected, untouched holdout set.

## Conclusion

The project involved challenges relating to missing values, unusual observations, skewed target values, feature engineering, correlated inputs, categorical encoding, model stability, validation, and buyer recommendations.

We addressed these issues using data inspection, appropriate imputation, feature engineering, IQR analysis, preprocessing pipelines, regularized regression, ensemble models, cross-validation, error analysis, and model explainability.

The tuned Gradient Boosting Regressor is our current primary candidate based on cross-validation MAE. However, its final suitability for production requires independent validation, additional error analysis, and verification that the data and model meet the intended business requirements.

```python
def recommend_houses(
    data,
    max_budget,
    min_area=0,
    min_bedrooms=1,
    min_bathrooms=1,
    min_quality=1,
    preferred_neighborhood=None
):
    # Start with the buyer's budget
    matches = data[data["SalePrice"] <= max_budget]

    # Apply property requirements
    matches = matches[
        (matches["GrLivArea"] >= min_area) &
        (matches["BedroomAbvGr"] >= min_bedrooms) &
        (matches["TotalBathrooms"] >= min_bathrooms) &
        (matches["OverallQual"] >= min_quality)
    ]

    # Apply neighborhood preference if provided
    if preferred_neighborhood is not None:
        matches = matches[
            matches["Neighborhood"].eq(preferred_neighborhood)
        ]

    # Display the most affordable matching properties first
    result_columns = [
        "Id",
        "Neighborhood",
        "SalePrice",
        "GrLivArea",
        "BedroomAbvGr",
        "TotalBathrooms",
        "OverallQual",
        "YearBuilt"
    ]

    return matches[result_columns].sort_values(
        "SalePrice",
        ascending=True
    )


# Example buyer requirements
recommendations = recommend_houses(
    data=df,
    max_budget=200000,
    min_area=1500,
    min_bedrooms=3,
    min_bathrooms=2,
    min_quality=6
)

print("Number of matching historical houses:", len(recommendations))
display(recommendations.head(10))
```

```python
import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.model_selection import KFold, cross_validate

# 1. Select the 11 user-friendly features
app_features = [
    "OverallQual",
    "OverallCond",
    "GrLivArea",
    "TotalBsmtSF",
    "GarageCars",
    "FullBath",
    "BedroomAbvGr",
    "YearBuilt",
    "LotArea",
    "Neighborhood",
    "KitchenQual"
]

X_app = model_df[app_features].copy()
y_app = model_df["SalePrice"].copy()

# 2. Separate numerical and categorical features
app_numeric_features = [
    "OverallQual",
    "OverallCond",
    "GrLivArea",
    "TotalBsmtSF",
    "GarageCars",
    "FullBath",
    "BedroomAbvGr",
    "YearBuilt",
    "LotArea"
]

app_categorical_features = [
    "Neighborhood",
    "KitchenQual"
]

# 3. Create preprocessing for numerical features
app_numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median"))
])

# 4. Create preprocessing for categorical features
app_categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# 5. Combine preprocessing steps
app_preprocessor = ColumnTransformer(
    transformers=[
        ("num", app_numeric_transformer, app_numeric_features),
        ("cat", app_categorical_transformer, app_categorical_features)
    ]
)

# 6. Create the complete prediction pipeline
app_prediction_model = Pipeline(steps=[
    ("preprocessor", app_preprocessor),
    ("model", GradientBoostingRegressor(
        n_estimators=300,
        learning_rate=0.1,
        max_depth=3,
        min_samples_leaf=2,
        subsample=0.8,
        random_state=42
    ))
])

# 7. Evaluate using five-fold cross-validation
cv = KFold(n_splits=5, shuffle=True, random_state=42)

scoring = {
    "MAE": "neg_mean_absolute_error",
    "RMSE": "neg_root_mean_squared_error",
    "R2": "r2"
}

cv_scores = cross_validate(
    app_prediction_model,
    X_app,
    y_app,
    cv=cv,
    scoring=scoring,
    n_jobs=1
)

print("Form-ready model: 5-fold CV results")
print(f"Mean MAE:  ${-cv_scores['test_MAE'].mean():,.2f}")
print(f"Mean RMSE: ${-cv_scores['test_RMSE'].mean():,.2f}")
print(f"Mean R²:    {cv_scores['test_R2'].mean():.4f}")

# 8. Fit the final application model on all historical records
app_prediction_model.fit(X_app, y_app)

print("\nApplication model trained successfully.")
print("Training records:", len(X_app))
print("Input features:", len(app_features))
```

```python
%%writefile /content/drive/MyDrive/PRCP_1020_HousePricePred/app.py

from pathlib import Path

import joblib
import numpy as np
import pandas as pd
import streamlit as st


# --------------------------------------------------
# 1. APP CONFIGURATION
# --------------------------------------------------

st.set_page_config(
    page_title="House Price Assistant",
    page_icon="🏠",
    layout="wide"
)

BASE_DIR = Path(__file__).resolve().parent

MODEL_PATH = BASE_DIR / "house_price_model.joblib"
RECOMMENDATION_PATH = BASE_DIR / "house_recommendation_data.csv"


# --------------------------------------------------
# 2. LOAD SAVED MODEL AND HISTORICAL DATA
# --------------------------------------------------

@st.cache_resource
def load_model(path):
    return joblib.load(path)


@st.cache_data
def load_houses(path):
    return pd.read_csv(path)


if not MODEL_PATH.exists():
    st.error("Model file not found: house_price_model.joblib")
    st.stop()

if not RECOMMENDATION_PATH.exists():
    st.error(
        "Recommendation data not found: "
        "house_recommendation_data.csv"
    )
    st.stop()

try:
    model = load_model(str(MODEL_PATH))
    houses = load_houses(str(RECOMMENDATION_PATH))
except Exception as exc:
    st.error(f"Could not load the application files: {exc}")
    st.stop()


required_columns = [
    "Id", "Neighborhood", "SalePrice", "GrLivArea",
    "BedroomAbvGr", "TotalBathrooms", "OverallQual",
    "YearBuilt", "GarageCars", "Fireplaces", "KitchenQual"
]

missing_columns = [
    col for col in required_columns
    if col not in houses.columns
]

if missing_columns:
    st.error(f"Missing required data columns: {missing_columns}")
    st.stop()


# Flag unusual size-price combinations for review.
# This rule identifies candidates for review, not confirmed errors.
houses["Review_Status"] = np.where(
    (houses["GrLivArea"] > 4000) &
    (houses["SalePrice"] < 300000),
    "Review required",
    "No flag under this rule"
)


neighborhood_options = sorted(
    houses["Neighborhood"].dropna().astype(str).unique().tolist()
)

kitchen_order = ["Po", "Fa", "TA", "Gd", "Ex"]

available_kitchens = set(
    houses["KitchenQual"].dropna().astype(str).unique()
)

kitchen_options = [
    quality for quality in kitchen_order
    if quality in available_kitchens
]

year_min = int(houses["YearBuilt"].min())
year_max = int(houses["YearBuilt"].max())


# --------------------------------------------------
# 3. APPLICATION HEADER
# --------------------------------------------------

st.title("🏠 House Price Assistant")

st.write(
    "Estimate a house price or discover historical properties "
    "that match your preferences."
)

st.info(
    "This prototype uses historical house sales from Ames, Iowa. "
    "Recommendations are not current listings, and predictions "
    "are estimates rather than professional appraisals."
)


# --------------------------------------------------
# 4. TWO APPLICATION TABS
# --------------------------------------------------

predict_tab, recommend_tab = st.tabs(
    ["💰 Predict House Price", "🔎 Find Matching Houses"]
)


# --------------------------------------------------
# 5. PRICE PREDICTION FORM
# --------------------------------------------------

with predict_tab:

    st.subheader("Enter the property characteristics")

    st.write(
        "Choose the available values that best describe the house. "
        "The trained model uses these 11 features."
    )

    with st.form("prediction_form"):

        col1, col2 = st.columns(2)

        with col1:
            overall_qual = st.slider(
                "Overall quality",
                min_value=1,
                max_value=10,
                value=6
            )

            overall_cond = st.slider(
                "Overall condition",
                min_value=1,
                max_value=10,
                value=5
            )

            gr_liv_area = st.number_input(
                "Above-ground living area (sq. ft.)",
                min_value=300,
                max_value=6000,
                value=1500,
                step=50
            )

            total_bsmt_sf = st.number_input(
                "Basement area (sq. ft.)",
                min_value=0,
                max_value=10000,
                value=800,
                step=50
            )

            garage_cars = st.selectbox(
                "Garage capacity (cars)",
                options=[0, 1, 2, 3, 4],
                index=2
            )

        with col2:
            full_bath = st.selectbox(
                "Full bathrooms above ground",
                options=[0, 1, 2, 3, 4],
                index=2
            )

            bedroom_abvgr = st.selectbox(
                "Bedrooms above ground",
                options=list(range(1, 9)),
                index=2
            )

            year_built = st.slider(
                "Year built",
                min_value=year_min,
                max_value=year_max,
                value=min(max(2000, year_min), year_max)
            )

            lot_area = st.number_input(
                "Lot area (sq. ft.)",
                min_value=1000,
                max_value=250000,
                value=8500,
                step=500
            )

            neighborhood = st.selectbox(
                "Neighborhood",
                options=neighborhood_options
            )

        kitchen_qual = st.selectbox(
            "Kitchen quality",
            options=kitchen_options,
            format_func=lambda x: {
                "Ex": "Excellent",
                "Gd": "Good",
                "TA": "Typical / Average",
                "Fa": "Fair",
                "Po": "Poor"
            }.get(x, x)
        )

        predict_clicked = st.form_submit_button(
            "Predict House Price",
            use_container_width=True
        )

    if predict_clicked:

        input_data = pd.DataFrame([{
            "OverallQual": overall_qual,
            "OverallCond": overall_cond,
            "GrLivArea": gr_liv_area,
            "TotalBsmtSF": total_bsmt_sf,
            "GarageCars": garage_cars,
            "FullBath": full_bath,
            "BedroomAbvGr": bedroom_abvgr,
            "YearBuilt": year_built,
            "LotArea": lot_area,
            "Neighborhood": neighborhood,
            "KitchenQual": kitchen_qual
        }])

        try:
            predicted_price = float(model.predict(input_data)[0])

            if not np.isfinite(predicted_price):
                st.error("The model returned an invalid price.")
            else:
                st.success("Price prediction completed.")

                st.metric(
                    "Estimated historical-market sale price",
                    f"${predicted_price:,.0f}"
                )

                st.caption(
                    "The estimate is based on historical training data. "
                    "It is not a guarantee of a property's current market value."
                )

                with st.expander("Review the inputs used"):
                    st.dataframe(
                        input_data,
                        use_container_width=True
                    )

        except Exception as exc:
            st.error(f"Could not predict the house price: {exc}")


# --------------------------------------------------
# 6. BUYER RECOMMENDATION FORM
# --------------------------------------------------

with recommend_tab:

    st.subheader("Find properties matching your requirements")

    st.write(
        "Filter historical sales by budget, property features, "
        "neighborhood, and optional preferences."
    )

    with st.form("recommendation_form"):

        budget_range = st.slider(
            "Sale-price budget range ($)",
            min_value=0,
            max_value=800000,
            value=(100000, 200000),
            step=5000
        )

        preferred_neighborhoods = st.multiselect(
            "Preferred neighborhoods (leave empty for all)",
            options=neighborhood_options,
            default=[]
        )

        col1, col2 = st.columns(2)

        with col1:
            min_area = st.number_input(
                "Minimum living area (sq. ft.)",
                min_value=0,
                max_value=6000,
                value=1500,
                step=100
            )

            min_bedrooms = st.selectbox(
                "Minimum bedrooms",
                options=list(range(0, 9)),
                index=3
            )

            min_bathrooms = st.selectbox(
                "Minimum total bathroom equivalents",
                options=[0.0, 0.5, 1.0, 1.5, 2.0,
                         2.5, 3.0, 3.5, 4.0, 4.5, 5.0],
                index=4
            )

        with col2:
            min_quality = st.slider(
                "Minimum overall quality",
                min_value=1,
                max_value=10,
                value=6
            )

            require_garage = st.checkbox(
                "Require a garage"
            )

            require_fireplace = st.checkbox(
                "Require a fireplace"
            )

            exclude_review = st.checkbox(
                "Hide records flagged for review",
                value=True
            )

        sort_preference = st.selectbox(
            "Rank matching houses by",
            options=[
                "Lowest historical price",
                "Highest quality",
                "Largest living area"
            ]
        )

        find_clicked = st.form_submit_button(
            "Find Matching Houses",
            use_container_width=True
        )

    if find_clicked:

        matches = houses[
            (houses["SalePrice"] >= budget_range[0]) &
            (houses["SalePrice"] <= budget_range[1]) &
            (houses["GrLivArea"] >= min_area) &
            (houses["BedroomAbvGr"] >= min_bedrooms) &
            (houses["TotalBathrooms"] >= min_bathrooms) &
            (houses["OverallQual"] >= min_quality)
        ].copy()

        if preferred_neighborhoods:
            matches = matches[
                matches["Neighborhood"].isin(
                    preferred_neighborhoods
                )
            ]

        if require_garage:
            matches = matches[matches["GarageCars"] >= 1]

        if require_fireplace:
            matches = matches[matches["Fireplaces"] >= 1]

        if exclude_review:
            matches = matches[
                matches["Review_Status"] == "No flag under this rule"
            ]

        if sort_preference == "Lowest historical price":
            matches = matches.sort_values("SalePrice")

        elif sort_preference == "Highest quality":
            matches = matches.sort_values(
                ["OverallQual", "SalePrice"],
                ascending=[False, True]
            )

        else:
            matches = matches.sort_values(
                ["GrLivArea", "SalePrice"],
                ascending=[False, True]
            )

        st.subheader("Recommendation results")

        st.write(f"Matching historical houses: **{len(matches)}**")

        result_columns = [
            "Id",
            "Neighborhood",
            "SalePrice",
            "GrLivArea",
            "BedroomAbvGr",
            "TotalBathrooms",
            "OverallQual",
            "YearBuilt",
            "Review_Status"
        ]

        if matches.empty:
            st.warning(
                "No houses matched every selected requirement. "
                "Try expanding the budget or relaxing one of the filters."
            )
        else:
            st.dataframe(
                matches[result_columns].head(50),
                use_container_width=True,
                hide_index=True
            )

            if len(matches) > 50:
                st.info(
                    "Showing the first 50 matching houses. "
                    "Adjust your filters to narrow the results."
                )

            st.caption(
                "These are historical sale records, not current "
                "property listings. A record without a review flag "
                "is not guaranteed to be error-free."
            )

            csv_data = matches[result_columns].to_csv(index=False)

            st.download_button(
                "Download matching houses as CSV",
                data=csv_data,
                file_name="matching_houses.csv",
                mime="text/csv"
            )
```
